```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
from io import StringIO

csv_data = """Student_ID,Name,Department,Study_Hours,Attendance,Marks
1,Aisha,CSE,5,92,88
2,Rahul,ECE,3,85,76
3,Sana,CSE,6,95,91
4,Arjun,ISE,4,88,82
5,Meera,CSE,7,96,94
6,Aditya,ECE,2,78,68
7,Fathima,ISE,5,90,85
8,Kiran,CSE,4,87,79
9,Ananya,ECE,6,93,89
10,Rohan,ISE,3,82,72
11,Zoya,CSE,8,98,96
12,Vikram,ECE,4,86,77
13,Diya,ISE,6,91,87
14,Sameer,CSE,5,89,83
15,Nisha,ECE,7,94,92"""

df = pd.read_csv(StringIO(csv_data))

print("Dataset loaded successfully!")
df
```

```python
# Display the first 5 rows
print("First 5 rows:")
display(df.head())

# Display the last 5 rows
print("\nLast 5 rows:")
display(df.tail())

# Display number of rows and columns
print("\nDataset Shape:")
print(df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display dataset information
print("\nDataset Information:")
df.info()

# Display statistical summary
print("\nStatistical Summary:")
display(df.describe())
```

```python
# Check for missing values
print("Missing Values:")
print(df.isnull().sum())

# Check for duplicate rows
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

# Remove duplicate rows if any
df = df.drop_duplicates()

print("\nDataset after removing duplicates:")
print(df.shape)
```

```python
# NumPy Analysis of Student Marks

marks = df["Marks"].values

# Calculate average marks
mean_marks = np.mean(marks)

# Find highest marks
highest_marks = np.max(marks)

# Find lowest marks
lowest_marks = np.min(marks)

# Calculate standard deviation
marks_std = np.std(marks)

print("Average Marks:", mean_marks)
print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)
print("Standard Deviation:", marks_std)
```

```python
# Pandas Data Analysis

# 1. Calculate average marks for each department
department_average = df.groupby("Department")["Marks"].mean()

print("Average Marks by Department:")
print(department_average)

# 2. Calculate average study hours
average_study_hours = df["Study_Hours"].mean()

print("\nAverage Study Hours:")
print(average_study_hours)

# 3. Find the student with the highest marks
highest_student = df.loc[df["Marks"].idxmax()]

print("\nStudent with Highest Marks:")
print(highest_student)

# 4. Find students who scored more than 80 marks
high_scorers = df[df["Marks"] > 80]

print("\nStudents Scoring Above 80:")
display(high_scorers)
```

```python
# Bar Chart - Average Marks by Department

department_average = df.groupby("Department")["Marks"].mean()

plt.figure(figsize=(8, 5))

plt.bar(
    department_average.index,
    department_average.values
)

plt.title("Average Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")

plt.show()
```

```python
# Line Chart - Student Marks

plt.figure(figsize=(10, 5))

plt.plot(
    df["Student_ID"],
    df["Marks"],
    marker="o"
)

plt.title("Student Marks")
plt.xlabel("Student ID")
plt.ylabel("Marks")

plt.grid(True)

plt.show()
```

```python
# Categorize students based on marks

def categorize_marks(mark):
    if mark >= 85:
        return "Excellent"
    elif mark >= 70:
        return "Good"
    else:
        return "Needs Improvement"


df["Performance"] = df["Marks"].apply(categorize_marks)

print("Student Performance Categories:")
display(df[["Name", "Marks", "Performance"]])
```

```python
# Pie Chart - Student Performance Distribution

performance_counts = df["Performance"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    performance_counts.values,
    labels=performance_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Performance Distribution")

plt.show()
```
